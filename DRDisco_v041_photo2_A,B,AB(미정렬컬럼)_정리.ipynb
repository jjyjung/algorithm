#### def read_file();

```python
# 파일 읽기 및 전처리
def read_file(file, X_list, time_list):
  df = pd.read_csv(file)
  # X 컬럼 목록 저장
  df_X = df[X_list]

  # 날짜 노이즈 수정
  df_X = df_X.replace('0001-01-01 00:00:00', '1900-01-01 00:00:00')

  # datatime 포맷으로 수정
  # time_list = ['AI_07_LOT_ESTIMATE_END_TIME','AI_08_LPST_DATE','AI_09_LOT_END_TIME']
  for col in time_list:
    df_X[col] = pd.to_datetime(df_X[col], format="%Y-%m-%d %H:%M:%S.%f")
    df_X[col] = df_X[col].apply(lambda x: x.value)

  # Min-Max Normalization
  # scaler = MinMaxScaler()
  # scaler.fit(df_X)
  # df_norm = scaler.transform(df_X)

  # 최대-최소 정규화 생략 (해석 용이성 때문)
  df_norm = df_X
  df_norm = pd.DataFrame(df_norm, columns=df_X.columns, index=list(df_X.index.values))

  return df, df_norm
```

## 분석 함수

```python
import pandas as pd

# 데이터프레임 예시
data = {
    'cnt_gt': [1, 0, 3, 2, 10],
    'cnt_lt': [0, 3, 1, 8, 0],
}
cnt = pd.DataFrame(data)

def extract_zero_cols_by_order(is_known_cols=True):
    if is_known_cols == True:
        return cnt.index
    else:
        filtered_df = cnt[
            (cnt['cnt_gt'] * cnt['cnt_lt'] == 0) &  # 곱이 0인 조건
            ((cnt['cnt_gt'] + cnt['cnt_lt']) > 0)   # 합이 양수인 조건
        ]

        # cnt_gt와 cnt_lt의 합으로 내림차순 정렬
        sorted_df = filtered_df.assign(sum=filtered_df['cnt_gt'] + filtered_df['cnt_lt']).sort_values(by='sum', ascending=False)

        # 인덱스 추출
        result_indices = sorted_df.index.tolist()
        print("조건에 맞는 인덱스:", result_indices)
        return result_indices

extract_zero_cols_by_order(False)

def extract_diff_cols_by_order(is_known_cols=True):
    if is_known_cols == True:
        return cnt.index
    else:
        filtered_df = cnt[
            (cnt['cnt_gt'] * cnt['cnt_lt'] > 0)   # 곱이 양인 조건
        ]
        # cnt_gt와 cnt_lt의 차이 계산 및 내림차순 정렬
        sorted_df = filtered_df.assign(diff=(filtered_df['cnt_gt'] - filtered_df['cnt_lt']).abs()).sort_values(by='diff', ascending=False)

        # 인덱스 추출
        result_indices = sorted_df.index.tolist()
        print("조건에 맞는 인덱스:", result_indices)
        return result_indices

extract_diff_cols_by_order(False)
```

```python
import re

def extract_variables_thresholds(condition_str):
    """조건문 문자열에서 변수명과 임계치를 추출합니다.

    Args:
    condition_str: 조건문을 포함한 문자열.

    Returns:
    변수명과 임계치를 포함한 튜플의 리스트.
    """
    # 변수명, 연산자, 임계치를 찾기 위한 정규 표현식
    pattern = r'(\w+)\s*(<=|>=|!=|==|<|>)\s*([-+]?\d*\.?\d+)'

    # 패턴과 일치하는 모든 부분 찾기
    matches = re.findall(pattern, condition_str)

    # 추출된 정보를 리스트로 반환
    return [(match[0], match[1], float(match[2])) for match in matches]

# 조건문 예시
condition_example = "AI_15_SCHED_STEPSEQ <= 1.50 AND AI_15_SCHED_STEPSEQ > -91.00"

# 함수 실행 및 결과 출력
extracted_info = extract_variables_thresholds(condition_example)
for info in extracted_info:
    print(info)
```

```python
def remove_trailing_zeros(number):
    # 숫자를 문자열로 변환
    num_str = str(number)
    # 소수점이 있으면 불필요한 0과 소수점 제거
    if '.' in num_str:
        num_str = num_str.rstrip('0').rstrip('.')
    return num_str

def convert_condition(condition):
  feature = condition[0]
  operator = condition[1]
  threshold = condition[2]
  if feature[:-2] in time_list:
    print("날짜 포맷이면 feature, threshold = ", feature, threshold)
    # np.float64에서 datetime 객체로 변환하기 위해 초 단위로 변환
    # 주의: np.float64 값이 나노초 단위로 되어 있다고 가정 (일반적인 pandas timestamp 나노초 단위)
    threshold_sec = threshold / 1e9  # 나노초를 초로 변환
    # datetime 객체로 변환
    threshold = datetime.utcfromtimestamp(threshold_sec).strftime('%Y-%m-%d %H:%M:%S')
    threshold = '#' + threshold + '#'

  elif feature[:-2] in int_list:
    # > 또는 <
    if operator == '>':
      threshold = math.ceil(threshold)
      operator = '>='
    elif operator == '<':
      threshold = math.floor(threshold)
      operator = '<='
    elif operator == '>=' and (threshold - int(threshold) != 0):
      threshold = math.ceil(threshold)
    elif operator == '=<' and (threshold - int(threshold) != 0):
      threshold = math.floor(threshold)

    # threshold 정수화
    if threshold - int(threshold) == 0:
      threshold = int(threshold)

  # AI_08_NEW_ARR_BOM_USING_RATIO_A <= 0.02500000037252903
  elif feature[:-2] in float_list:
    print("실수형 처리 진행...")
    if (threshold - round(threshold, 5) == threshold - round(threshold, 8)):
      print("실수형 처리 진행...")
      threshold_str = remove_trailing_zeros(round(threshold, 5))
      threshold = float(threshold_str)
      if operator == '>=':
        operator = '>'
      elif operator == '<':
        operator = '<='
      elif operator == '<=':
        pass
      elif operator == '>':
        pass

  llist = list(condition)
  llist[0] = feature
  llist[1] = operator
  llist[2] = threshold
  return tuple(llist)

def postprocess_cond(cond):
  # cond 파싱
  conds = extract_variables_thresholds(cond)

  cond_str = ''
  for i, condition in enumerate(conds):
    conds[i] = convert_condition(condition)

    # 조건문 연결 작성
    for con in conds[i]:
      cond_str += str(con) + " "
    cond_str += 'AND '

  print('cond_str =', cond_str)
  cond = cond_str[:-5]

  return cond
```
